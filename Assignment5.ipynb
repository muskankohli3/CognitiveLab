```python
#Q4
muskan = np.linspace(10,100,25)
print(muskan.ndim)
print(muskan.shape)
print(muskan.size)
print(muskan.dtype)
print(muskan.nbytes)
transpose1 = muskan.reshape(muskan.size,1)
print(transpose1)
transpose2 = muskan.T
print(transpose2)
```
