```python
#Q3(i)
import random
random.seed(1024170453) 
randL = [random.randint(100,900) for _ in range(100)]
print(randL)
```

```python
#Q3(ii)
oddL = [l for l in randL if l%2!=0]
print(oddL)
print(len(oddL))
```

```python
#Q3(iii)
eveL = [l for l in randL if l%2==0]
print(eveL)
print(len(eveL))
```

```python
#Q3(iv)
def primeN(num):
    for i in range (2,num//2+1):
        if num%i==0:
            return 0
    return 1

primeL = [l for l in randL if primeN(l)==True]
print(primeL)
print(len(primeL))
```

```python
#Q3(v)
from collections import Counter

freq = Counter(randL)
most_common_num, frequency = freq.most_common(1)[0]

print("Most frequent number:", most_common_num)
print("Number of times it occurs:", frequency)
```

```python
#Q5(i)
my_dict = {
    "name": "Muskan Kohli",
    "roll_no": "1024170453",
    "branch": "CSE",
    "age": 20,
    "city": "Patiala"
}
print(my_dict)
my_dict["location"] = my_dict.pop("city")
print(my_dict)
```

```python
#Q5(ii)
my_dict["cgpa"] = 9.91
print(my_dict)
```

```python
#Q5(iii)
my_dict["age"] += 1
print(my_dict)
```

```python
#Q5(iv)
dict_pop = my_dict.copy()
dict_del = my_dict.copy()

dict_pop.pop("branch")
del dict_del["branch"]

print(dict_pop)
print(dict_del)

# pop() removes the key and returns its value, whereas del removes the key without returning its value.
```

```python
#Q5(v)
for key, value in my_dict.items():
    print(key, "->", value)
```

```python
#Q5(vi)
if "email" in my_dict:
    print("Email:", my_dict["email"])
else:
    print("Email key does not exist.")
```

```python
#Q5(vii)
friend_dict = {
    "name": "Aman",
    "roll_no": "1024170001",
    "branch": "ECE",
    "age": 21,
    "city": "Chandigarh"
}

merged_dict = {**my_dict, **friend_dict}
print(merged_dict)

# When both dictionaries have the same key, the value from the second dictionary (friend_dict) wins.
```
