```python
#Q4(a)
X = np.array([[6, 70, 3],
              [5, 50, 6],
              [8, 80, 2],
              [4, 30, 8]])
y = np.array([40, 65, 30, 85])
print(X.shape)
print(X.ndim)
```

```python
#Q4(b)
print(X.T)
print("Transpose represents features as rows and users as columns.")
```

```python
#Q4(c)
print(X.T @ X)
```

```python
#Q4(d)
inverse = np.linalg.inv(X.T @ X)
print(inverse)
```

```python
#Q4(e)
beta = inverse @ X.T @ y
print(beta)
```

```python
#Q4(f)
print("The coefficients represent the contribution of sleep hours, activity level, and stress level to the predicted assistance score.")
```

```python
#Q4(g)
new_user = np.array([5, 40, 7])
predicted_score = new_user @ beta
print(predicted_score)
```
